# Example: Summing the 'price' column from a gzip CSV file using PySpark
This notebook creates a sample gzip CSV file with a `price` column, sums the values in this column using PySpark, and displays the result.

```python
import pandas as pd
import gzip
import os

# 1. Create a sample DataFrame with a 'price' column
data = {'price': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)

# 2. Save the DataFrame to a gzip CSV file
csv_gz_path = '/tmp/example_prices.csv.gz'
df.to_csv(csv_gz_path, index=False, compression='gzip')

print(f'Gzip CSV file saved: {csv_gz_path}')
```

```python
# 3. Sum the 'price' column using PySpark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("SumCSVPrice").master("local[*]").getOrCreate()

# Read the gzip CSV file
df_spark = spark.read.option("header", True).csv(csv_gz_path)

# Ensure the 'price' column is numeric
from pyspark.sql.functions import col

df_spark = df_spark.withColumn("price", col("price").cast("double"))

# Calculate the sum
sum_row = df_spark.agg({"price": "sum"}).collect()[0]
sum_price = sum_row[0]
```

```python
# 4. Display the sum
print(f"Sum of 'price' column: {sum_price}")
```
